```python
#HYPOTHESIS TESTING


#1. State the Hypotheses Statement
#Null Hypothesis (H₀): The weekly operating cost model is accurate.
#Alternative Hypothesis (H₁): The weekly operating cost model is not accurate, implying that the weekly operating cost is higher.


import math
import scipy.stats as stats

# Given data
sample_mean = 3050  # Rs.
theoretical_mean = 4000  # Rs.
sigma = 125  # Rs.
n = 25  # Sample size

# Calculate the test statistic
t = (sample_mean - theoretical_mean) / (sigma / math.sqrt(n))
print(f"Test Statistic (t): {t}")

# Determine the critical value for a one-tailed test with alpha = 0.05
alpha = 0.05
critical_value = stats.norm.ppf(1 - alpha)
print(f"Critical Value (Z): {critical_value}")

# Make a decision
if t > critical_value:
    decision = "Reject the null hypothesis"
else:
    decision = "Fail to reject the null hypothesis"

print(f"Decision: {decision}")
```

```python
print("Conclusion:")
print("Based on the analysis, there is no strong evidence to support the restaurant owners\n claim that the weekly operating costs are higher than the model suggests.")
```

```python
#CHISQUARE TESTING


#State the Hypotheses:
#Null Hypothesis (H₀): There is no association between the type of smart home device purchased and the customer satisfaction level.
#Alternative Hypothesis (H₁): There is a significant association between the type of smart home device purchased and the customer satisfaction level.


import numpy as np
import scipy.stats as stats

# Observed frequencies
observed = np.array([
    [50, 70],
    [80, 100],
    [60, 90],
    [30, 50],
    [20, 50]
])

# Row and column totals
row_totals = observed.sum(axis=1)
col_totals = observed.sum(axis=0)
total = observed.sum()

# Expected frequencies
expected = np.outer(row_totals, col_totals) / total

# Compute the Chi-Square statistic
chi_square_stat = ((observed - expected) ** 2 / expected).sum()
print(f"Chi-Square Statistic (χ²): {chi_square_stat}")

# Degrees of freedom
df = (observed.shape[0] - 1) * (observed.shape[1] - 1)
print(f"Degrees of Freedom (df): {df}")

# Critical value for alpha = 0.05
alpha = 0.05
critical_value = stats.chi2.ppf(1 - alpha, df)
print(f"Critical Value (χ² critical): {critical_value}")

# Make a decision
if chi_square_stat > critical_value:
    decision = "Reject the null hypothesis"
else:
    decision = "Fail to reject the null hypothesis"

print(f"Decision: {decision}")
```

```python
print("Conclusion:")
print("Based on the analysis, there is no significant association between the type of smart home device purchased \n (Smart Thermostats vs. Smart Lights) and the customer satisfaction level.")
```
